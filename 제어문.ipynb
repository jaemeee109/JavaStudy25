```python
# 비교 연산자
# x > y : x가 y보다 크다
# x < y : x가 y보다 작다
# x == y : x와 y 가 같다
# x != y : x와 y가 같지 않다
# >= y : x가 y보다 크거나 같다
# x <= y : x가 y보다 작거나 같다
x = 3
y = 2
print(x > y)
print(x < y)
print(x == y)
print(x != y)
```

```python
# 만약 3,000원 이상의 돈을 가지고 있으면 택시를 타고 가고,
# 그렇지 않으면 걸어가라
money = 2000
if money >= 3000 :
  print("택시를 타고 가라")
else:
  print("걸어가라")
```

```python
# 조건을 판단하기 위해 사용하는 다른 연산자 and , or, not
# x or y : x와 y 둘 중 하나만 참이어도 참이다
# x and y : x와 y 모두 참이어야 참이다
# not x : x가 거짓이면 참이다
# =======================================================
# 만약 3,000원 이상의 돈을 가지고 있으면 택시를 타고 가고,
# 그렇지 않으면 걸어가라
money = 2000
card = True
if money >= 3000 or card:
  print("택시를 타고 가라")
else:
   print("걸어가라")
# money는 2,000원 이지만, card가 True 이기 때문에 조건문이 참이 됨
```

```python
# in, not in
# x in 리스트 : x not in 리스트
# x in 튜플 : x not in 튜플
# x in 문자열 : x not in 문자열
#====================================
# (1) 리스트
# 1이 [1,2,3] 안에 있으므로 참
print(1 in [1,2,3])
# 1이 [1,2,3] 안에 없냐는 조건인데 있으므로 거짓
print(1 not in [1,2,3])
#====================================
# (2) 튜플
print('a' in ('a','b','c'))
print('j' not in 'python')
#====================================
# (3) 문자열
# 만약 주머니에 돈이 있으면 택시를 타고 가고, 없으면 걸어가라.
pocket = ['paper', 'cellphone','money']
if 'money' in pocket:
  print("택시를 타고 가라")
else:
  print("걸어가라")
```

```python
# 조건문에서 아무것도 실행되지 않게 설정할 때 pass
# 주머니에 돈이 있으면 가만히 있고, 주머니에 돈이 없으면 카드를 꺼내라.
pocket = ['paper', 'cellphone','money']
if 'money' in pocket:
  pass
else:
  print("카드를 꺼내라")
# pocket 안에 money 문자열이 있기 때문에 pass가 수행되고
# 결과값이 나오지 않음
```

```python
# 다양한 조건을 판단하는 elif
# ===========================================
# 주머니에 돈이 있으면 택시를 타고 가고,
# 주머니에 돈은 없지만 카드가 있으면 택시를 타고 가고,
# 돈도 없고 카드도 없으면 걸어가라.
# ===========================================
# (1) if와 else만으로 표현
pocket = ['paper', 'cellphone']
card = True
if 'money' in pocket:
  print ("택시를 타고 가라")
else:
  if card :
    print ("택시를 타고 가라")
  else:
    print("걸어가라")
# ===========================================
# (2) elif를 사용해서 표현
if 'money' in pocket:
  print("택시를 타고 가라")
elif card:
  print("택시를 타고 가라")
else:
  print("걸어가라")
```

# while문
<b>while 조건문:</b>
<p/>&nbsp;&nbsp;&nbsp;&nbsp;수행할 문장 1
<p/>&nbsp;&nbsp;&nbsp;&nbsp;수행할 문장 2
<p/>&nbsp;&nbsp;&nbsp;&nbsp;수행할 문장 3


```python
treeHit = 0
while treeHit < 10:
  treeHit = treeHit + 1
  print("나무를 %d번 찍었습니다." % treeHit)
  if treeHit == 10:
    print ("나무 넘어갑니다")
```

```python
# while문 만들기
# ---------------
prompt = """
  1. Add
  2. Del
  3. List
  4. Quit
  Enter number: """
# ---------------
number=0 # 입력 받기 전 , 임시설정
while number != 4:
  print(prompt) # 4가 아닐동안 프롬프트 출력
  number = int(input()) # 사용자가 입력한 값을 int로 바꿔서
                        # number 변수에 저장
                        # 사용자가 4를 입력하면 False되어 반복종료

```

```python
# while문 강제로 빠져나가기 break
coffe = 10
money = 300
while money:
  print ("돈을 받았으니 커피를 줍니다")
  coffe = coffe -1
  print("남은 커피의 양은 %d개 입니다" % coffe)
  print("-------------------------------------")
  if coffe == 0:
    print ("커피가 다 떨어졌습니다. 판매를 중지합니다.")
    break
```

```python
# 자판기
coffee = 10
while True:
  money = int(input("돈을 넣어 주세요: "))
  if money == 300 :
    print("커피를 줍니다.")
    coffee = coffee - 1
  elif money > 300:
    print("거스름돈 %d를 주고 커피를 줍니다." % (money-300))
    coffee = coffee - 1
  else :
    print ("돈을 다시 돌려주고 커피를 주지 않습니다")
    print ("남은 커피의 양은 %d개 입니다." % coffee)
  if coffee == 0:
    print("커피가 다 떨어졌습니다. 판매를 중지합니다")
    break
```

```python
# while문의 맨 처음으로 돌아가기 continue
# while문을 빠져나가지 않고 처음으로 돌아가게 하고 싶을 때
a = 0
while a < 10 : # a가 10보다 작을 동안 계속 반복해라
  a = a + 1 # 반복할 때마다 a를 1씩 증가시킴
  if a % 2 == 0: continue # a가 짝수라면 참(True)
                          # continue는 이번 반복을 건너뛰고,
                          # 다음 반복으로 넘어가라는 명령
  print(a) # a가 홀수일 때 실행 됨
```

```python
# while-else 문
count = 0
while count <3 :
  print(f"count: {count}")
  count += 1
else:
   print("while 문이 정상 종료되었습니다")
```

```python
# while-else 문에  break문 사용
# break를 사용해 while문을 빠져나가면 else는 실행되지 않음
count = 0
while count <5 :
  if count == 2:
    break
  print(f"count: {count}")
  count += 1
else:
   print("while 문이 정상 종료되었습니다")
```

```python
# 중첩된 while문
i = 1
while i <= 3: # i가 1~3 일 동안 반복
  j = 1 # j가 1~3일 동안 반복
  while j <= 3:
    print(f"i={i},j={j}")
    j += 1
  i += 1
```

# for 문
<b>for문의 기본 구조</b>
<p/>for 변수 in 리스트 (또는 튜플, 문자열):
<p/>&nbsp;&nbsp;&nbsp;&nbsp;수행할문장1
<p/>&nbsp;&nbsp;&nbsp;&nbsp;수행할문장2

```python
# 전형적인 for문
test_list = ['one','two','three']
for i in test_list:
  print(1)
```

```python
# 다양한 for문 사용
a = [(1,2),(3,4),(5,6)]
for(first, last) in a:
  print(first + last)
```

```python
# for문의 응용
# 총 5명의 학생이 시험을 보았는데 시험 점수가 60점 이상이면 합격이고
# 그렇지 않으면 불합격이다. 합격인지, 불합격인지 결과를 보여 주시오.
marks = [90,25,67,45,80] # 점수리스트
number = 0 # 학생에게 붙일 번호
for mark in marks:
  number = number +1
  if mark >=60:
    print("%d번 학생은 합격입니다" % number)
  else :
    print("%d 학생은 불합격 입니다" % number)
```

```python
# for문과 continue문
# 문장을 수행하는 도중 continue 문을 만나면 for 문의 처음으로 돌아가게 된다
marks = [90, 25, 67, 45, 80]

number = 0
for mark in marks:
    number = number +1
    if mark < 60:
        continue
    print("%d번 학생 축하합니다. 합격입니다. " % number)
```

```python
# for문과 자주 사용하는 range 함수
# range 함수는 숫자 리스트를 자동으로 만들어 준다
# a에 포함되는 0부터 10미만의 숫자를 포함하는 range 객체를 만들어 줌
a = range(10)
print (a)
# 시작 숫자오 끝 숫자를 지정하려면
# range(시작숫자, 끝숫자) 형태를 사용, 끝 숫자는 포함되지 않음
b = range(1,11)
print(b)
```

```python
# range 함수
add = 0
for i in range(1,11):
  add = add + i
print(add)
# 1부터 10까지 차례로 대입되어 add = add + i를 반복적으로 수행
```

```python
# range 함수 문장출력
marks = [90, 25, 67, 45, 80]
for number in range(len(marks)): # len(marks)는 5가 됨
  if marks[number] < 60:
    continue
  print ("%d번 학생 축하합니다. 합격입니다." % (number+1))
```

```python
# for문과 range를 이요한 구구단
for i in range(2,10):        # 구구단의 단(2~9)
     for j in range(1, 10):   # 곱하는 수 (1~9)
         print(i*j, end=" ") # end의 " "는 띄어쓰기용
     print('')
```
